```python
import random

n = [100,1000,10000,100000,1000000,10000000]

def Estpi(n):
	inside = 0
	for _ in range(n):
	    x = random.random()
	    y = random.random()
	    if x*x + y*y <= 1:
	        inside += 1
	return 4*(inside/n)

for i in n:
    print(i, Estpi(i))
```

```python
import matplotlib.pyplot as plt

xs = [random.random() for _ in range(20000)]
ys = [random.random() for _ in range(20000)]
colours = ["green" if x*x + y*y <= 1 else "red" for x, y in zip(xs, ys)]

plt.figure(figsize=(5, 5))
plt.scatter(xs, ys, c=colours, s=5)
plt.show()
```

```python
import numpy
import matplotlib.pyplot as plt

avg = []
for i in n:
    errors = []
    for j in range(20):
        error = Estpi(i) - numpy.pi
        errors.append(abs(error))
    mean = sum(errors) / len(errors)
    avg.append(mean)

plt.xscale("log")
plt.yscale("log")
plt.scatter(n, avg)
plt.show()
```
